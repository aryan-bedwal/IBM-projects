```python
t=t.drop(columns=['deck'])
```

```python
t['age'].fillna(t['age'].median(),inplace=True)
```

```python
t['embarked'].fillna(t['embarked'].mode()[0],inplace=True)
```

```python
t=t.dropna()
```

```python
grouped = t.groupby(['sex', 'pclass'])['survived'].mean().reset_index()

# Plot
sns.barplot(data=grouped, x='sex', y='survived', hue='pclass', palette='Set2')

# Labels and title
plt.title('Survival Rate by Sex and Pclass')
plt.ylabel('Survival Rate')
plt.xlabel('Sex')
plt.ylim(0, 1)
plt.legend(title='Pclass')
plt.show()
```

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[column] < lower) | (data[column] > upper)]

# Example: Outliers in fare
fare_outliers = detect_outliers_iqr(t, 'age')
print(fare_outliers[['age']])
```
